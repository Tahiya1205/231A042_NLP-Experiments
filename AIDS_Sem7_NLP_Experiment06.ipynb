```python
import nltk
from nltk.tokenize import word_tokenize
from nltk.tag import UnigramTagger, BigramTagger, pos_tag
from nltk.corpus import treebank
from nltk import pos_tag_sents
```

```python
nltk.download('averaged_perceptron_tagger')
nltk.download('punkt')
nltk.download('treebank')
```

```python
import nltk
nltk.download('punkt_tab')

sample_text = "The cat sleeps on the warm sofa while the dog watches quitely"

tokens = word_tokenize(sample_text)
```

```python
import nltk
nltk.download('averaged_perceptron_tagger_eng')
rule_based_tags = pos_tag(tokens)
```

```python
print(rule_based_tags)
```

```python
import nltk
nltk.download('tagsets_json')
nltk.help.upenn_tagset("NNS")
```

```python
import nltk
nltk.download('tagsets_json')
nltk.help.upenn_tagset("IN")
```

```python
train_sents = treebank.tagged_sents()

unigram_tagger = UnigramTagger(train_sents)
unigram_tags = unigram_tagger.tag(tokens)

bigram_tagger = BigramTagger(train_sents, backoff=unigram_tagger)
bigram_tags = bigram_tagger.tag(tokens)
```

```python
print("Rule-Based Tags:", rule_based_tags)
print("\nUnigram Tags:", unigram_tags)
print("\nBigram Tags:", bigram_tags)
```

```python
import pandas as pd

comparison_data = []
for i, token in enumerate(tokens):
    comparison_data.append({
        'Word': token,
        'Rule-Based Tag': rule_based_tags[i][1] if i < len(rule_based_tags) else None,
        'Unigram Tag': unigram_tags[i][1] if i < len(unigram_tags) else None,
        'Bigram Tag': bigram_tags[i][1] if i < len(bigram_tags) else None
    })

comparison_df = pd.DataFrame(comparison_data)
display(comparison_df)
```

```python
test_sents = treebank.tagged_sents()[3000:]
unigram_accuracy = unigram_tagger.evaluate(test_sents)
bigram_accuracy = bigram_tagger.evaluate(test_sents)

print(f"\nUnigram Tagger Accuracy: {unigram_accuracy * 100:.2f}%")
print(f"Bigram Tagger Accuracy: {bigram_accuracy * 100:.2f}%")
```
