```python
corpus = [
    "The cat sat on the mat.",
    "The dog saw the cat.",
    "The mat was under the cat.",
    "The birds sang a sweet song.",
    "The song was very sweet and clear."
]
vectorizer = TfidfVectorizer(stop_words='english')
X = vectorizer.fit_transform(corpus)

print(X.shape)
```

```python
tdidf_df = pd.DataFrame(
    X.toarray(),
    columns=vectorizer.get_feature_names_out(),
    index=[f"D{i+1}" for i in range(len(corpus))]
)
tdidf_df.round(2)
```

```python
print(X)
```

```python
print(vectorizer.get_feature_names_out())
```

```python
similarity_matrix = cosine_similarity(X)

similarity_matrix
```

```python
highest_similarity = -1
most_similar_pair = None

for i in range(len(corpus)):
  for j in range(i+1, len(corpus)):
    similarity = similarity_matrix[i][j]
    if similarity > highest_similarity:
      highest_similarity = similarity
      most_similar_pair = (i, j)

print(
    f"Most similar documents:"
    f"D{most_similar_pair[0]+1} and D{most_similar_pair[1]+1}"
    f"with a similarity of {highest_similarity:.2f}"
)
```

```python
def similar(D1,D2):
  vector=vectorizer.fit_transform([D1,D2])
  similarity=cosine_similarity(vector)
  return similarity
```
